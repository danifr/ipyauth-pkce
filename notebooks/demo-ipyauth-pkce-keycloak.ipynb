# ipyauth-pkce — Keycloak Demo

Authenticate via Keycloak and use the access token to call a protected API.

```python
from ipyauth_pkce import Auth, ParamsKeycloak
```

## Configure & Sign In

Update the parameters below to match your Keycloak setup.

```python
params = ParamsKeycloak(
    base_url='https://your-keycloak-server/auth',
    realm='your-realm',
    client_id='your-client-id',
    redirect_uri='http://localhost:8888/callback/',
    scope='openid profile',
    response_type='code',
)

session = Auth(params=params)
session
```

## Use the Token

After clicking "Sign In" above, the access token is available:

```python
print(session)
print('Authenticated:', session.is_authenticated)
print('Token (first 50 chars):', session.access_token[:50], '...')
```

## Call a Protected API

```python
import requests

# session.headers returns {'Authorization': 'Bearer eyJ...'}
r = requests.get('https://your-api.example.com/data', headers=session.headers)
print(r.status_code)
r.json()
```
