# ipyauth-pkce — Google Demo

Authenticate via Google and use the access token to call Google APIs.

```python
from ipyauth_pkce import Auth, ParamsGoogle
```

## Configure & Sign In

Update the parameters below to match your Google Cloud Console setup.

```python
params = ParamsGoogle(
    response_type='code',
    client_id='your-client-id.apps.googleusercontent.com',
    redirect_uri='http://localhost:8888/callback/',
    scope='profile openid',
)

session = Auth(params=params)
session
```

## Use the Token

```python
print(session)
print('Authenticated:', session.is_authenticated)
print('Token (first 50 chars):', session.access_token[:50], '...')
```

## Call a Google API

```python
import requests

r = requests.get('https://www.googleapis.com/oauth2/v1/userinfo', headers=session.headers)
print(r.status_code)
r.json()
```
