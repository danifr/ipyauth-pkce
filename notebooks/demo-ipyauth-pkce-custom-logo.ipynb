# ipyauth-pkce — Custom Logo Demo

Shows the new `logo=` override on `Auth`. No real auth server is needed to *see* the logo: the widget renders as soon as it is displayed.

Three variants below: inline SVG bytes, an SVG file path, and the default (built-in provider logo).

```python
from ipyauth_pkce import Auth, ParamsKeycloak

# Dummy params — only used to build the widget; no network call happens on display.
params = ParamsKeycloak(
    base_url='https://kc.example.com', realm='demo', client_id='demo',
    redirect_uri='http://localhost:8888/callback/', scope='openid profile',
)
```

## 1. Custom logo from inline SVG bytes (letters)
A downstream library can ship its logo and pass the raw bytes.

```python
svg = (
    b'<svg xmlns="http://www.w3.org/2000/svg" width="120" height="26">'
    b'<rect width="120" height="26" rx="4" fill="#2a9d8f"/>'
    b'<text x="60" y="18" font-family="Arial" font-size="14" '
    b'font-weight="bold" fill="#fff" text-anchor="middle">MYCO</text>'
    b'</svg>'
)
Auth(params=params, logo=svg)
```

## 2. Custom logo from an SVG file path
Points at the example SVG shipped in the package `img/` folder.

```python
from pathlib import Path
import ipyauth_pkce.ipyauth_widget as w

logo_path = Path(w.__file__).parent / 'img' / 'example-custom-logo.svg'
print('logo:', logo_path, '->', logo_path.exists())
Auth(params=params, logo=str(logo_path))
```

## 3. Default (no override) — built-in Keycloak logo

```python
Auth(params=params)
```
