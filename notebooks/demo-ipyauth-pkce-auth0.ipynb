# ipyauth-pkce — Auth0 Demo

Authenticate via Auth0 and use the access token to call a protected API.

```python
from ipyauth_pkce import Auth, ParamsAuth0
```

## Configure & Sign In

Update the parameters below to match your Auth0 setup.

```python
params = ParamsAuth0(
    response_type='code',
    domain='your-tenant.auth0.com',
    client_id='your-client-id',
    audience='your-api-audience',
    redirect_uri='http://localhost:8888/callback/',
    scope='profile openid',
)

session = Auth(params=params)
session
```

## Use the Token

```python
print(session)
print('Authenticated:', session.is_authenticated)
print('Token (first 50 chars):', session.access_token[:50], '...')
```

## Call a Protected API

```python
import requests

r = requests.get('https://your-api.example.com/data', headers=session.headers)
print(r.status_code)
r.json()
```
